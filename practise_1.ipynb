```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


data = {
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rudra"],
    "Gender": ["Male", "Female", "Male", "Female", "Male"],
    "Survived": [0, 1, 0, 1, 1]
}

result = pd.DataFrame(data)
survival_data = result.groupby("Gender")["Survived"].mean()
print(survival_data)

sns.barplot(x="Gender", y="Survived", data=result, palette="pastel")
plt.title("Survival Rate by Gender")
plt.ylabel("Survival Probability")
plt.show()
```
